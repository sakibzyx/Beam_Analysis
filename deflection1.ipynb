```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

<p>Determine the maximum deflection of a cantilever beam subjected to a UDL using Python.</p>


```python
# Data
w = 10.0 # N/m
l = 5.0 # m
e = 200e9 #Pa
i = 1e-6 #[m^4]

def cantiliver_beam(u,x):
    u1, u2 = u # u1 = y, u2 = dy/dx
    du1dx = u2
    du2dx = (-(w*x**2)/2 + w*x*l -(w*l**2)/2)/(e*i)
    return du1dx, du2dx
initial_condition = [0,0]
x_value = np.linspace(0,l,100)

sol = sp.integrate.odeint(cantiliver_beam, initial_condition, x_value)
deflection = sol.T[0]

print(min(deflection))
```

```python
# plot the deflection

plt.figure(figsize=(10,6))
plt.plot(x_value, deflection, color = 'red', label = 'Deflection')
plt.title('Deflection')
plt.xlabel('Span of the beam(m)')
plt.ylabel('Deflection')
plt.legend()
plt.grid(True)
plt.show()
```
