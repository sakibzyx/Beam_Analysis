```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

```python
# Data
p = 10.0   # N (Point load at free end)
l = 5.0    # m
e = 200e9  # Pa
i = 1e-6   # m^4

def cant_beam(u,x):
    u1, u2 = u
    du1dx = u2
    du2dx = (-p*(l-x))/(e*i)
    return du1dx, du2dx
initial_condition = [0,0]
x_value = np.linspace(0,l,100)

sol = sp.integrate.odeint(cant_beam, initial_condition, x_value)
deflection = sol.T[0]
rotation = sol.T[1]
print(min(deflection))
print(min(rotation))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

# deflection
plt.subplot(2, 1, 1)
plt.plot(x_value, deflection, color="b", linewidth=2, label="Deflection (y)")
plt.axhline(0, color="black", linestyle="--", linewidth=0.8)
plt.ylabel("Deflection (m)")  
plt.title("Deflection Profile along the Beam")  
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()

# rotation
plt.subplot(2, 1, 2)  
plt.plot(
    x_value, rotation, color="r", linewidth=2, label="Rotation (dy/dx / theta)"
)
plt.axhline(0, color="black", linestyle="--", linewidth=0.8)
plt.xlabel("Beam Length x (m)") 
plt.ylabel("Rotation (rad)") 
plt.title("Slope / Rotation Profile along the Beam") 
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```
