```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

<p>A simply supported beam has a total length of 10 ft and carries a 50 lb point load at its center. Using Python, calculate the support reactions, shear force, and bending moment along the beam.</p>

```python
Adist = []
Asf = []
Abm = []

def sf_bm():
    p = 50
    l = 10
    ra = 25
    rb = 25

    for i in np.linspace(0, l, l+1, dtype = int):
        Adist.append(i)

        #sf
        sf = ra
        if i >= 5:
            sf -= p
        Asf.append(sf)

        #bm
        bm = ra*i 
        if i >= 5:
            bm -= p*(i-5)
        Abm.append(bm)

    return Adist, Asf, Abm

result = sf_bm()

print(Adist)
print(Asf)
print(Abm)
```

```python
plt.figure(figsize=(10,8))

#sfd
plt.subplot(2,1,1)
plt.plot(Adist, Asf, color = 'blue', marker = 'o')
plt.fill_between(Adist, Asf, 0, color = 'blue', alpha = 0.2)
plt.axhline(0, color = 'black')
plt.title('sfd') 

#sfd
plt.subplot(2,1,2)
plt.plot(Adist, Abm, color = 'red', marker = 'o')
plt.fill_between(Adist, Abm, 0, color = 'red', alpha = 0.2)
plt.axhline(0, color = 'black')
plt.title('bmd')

plt.tight_layout
plt.show()
```

<p>A simply supported beam has a total length of 10 ft and carries two point loads of 50 lb at 3 ft and 70 lb at 7 ft from the left support. Using Python, calculate the support reactions, shear force, and bending moment along the beam.</p>

```python
Adist = []
Asf = []
Abm = []

def sf_bm_2p():
    # data
    l = 10
    p1 = 50 # 3ft
    p2 = 70 # 7 ft
    ra = 56
    rb = 64
    for i in np.linspace(0, l, l+1, dtype = int):
        Adist.append(i)

        #sf 
        sf = ra
        if i >= 3:
            sf -= p1
        if i >= 7:
            sf -= p2
        Asf.append(sf)

        #bm
        bm = ra * i
        if i >= 3 :
            bm -= p1*(i-3)
        if i >= 7 :
            bm -= p2 * (i-7)
        Abm.append(bm)

    return Adist, Asf, Abm

result = sf_bm_2p()

print(Adist)
print(Asf)
print(Abm)

```

```python
plt.figure(figsize=(10,8))
plt.subplot(2,1,1)

# SFD
plt.plot(Adist, Asf, color = 'blue', marker = 'o')
plt.fill_between(Adist, Asf, 0, color = 'blue', alpha = 0.2)
plt.axhline(0, color = 'black')
plt.xlabel('Distance (ft)')
plt.ylabel('Shear Force (lbs)')
plt.title('Shear Force Diagram (SFD)')
plt.grid(True)

# BMD
plt.subplot(2,1,2)
plt.plot(Adist, Abm, color = 'red', marker = 'o')
plt.fill_between(Adist, Abm, 0, color = 'red', alpha = 0.2)
plt.axhline(0, color = 'black')
plt.xlabel('Distance (ft)')
plt.ylabel('Bending Moment (lb-ft)')
plt.title('Bending Moment Diagram (BMD)')
plt.grid(True)

plt.tight_layout()
plt.show()
```
